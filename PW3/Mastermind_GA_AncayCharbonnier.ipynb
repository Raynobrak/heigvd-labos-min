## <font color="brown">🎳&nbsp;&nbsp;Define Game</font>

```python
import string
ALPHABET = list(string.ascii_uppercase) + [' ']

N_CHARS = len(ALPHABET)  # Number of characters in our alphabet (27)

TARGET_SENTENCE = "METHINKS IT IS LIKE A WEASEL"
N_PEGS = len(TARGET_SENTENCE) # Number of pegs in our genetic algorithm
```

```python
import numpy as np

# Function to display the board corresponding to a given chromosome
def display_board(chromosome):
    print(''.join([ALPHABET[i] for i in chromosome]))
```

```python
# See all the characters in the alphabet
display_board([i for i in np.arange(N_CHARS)])
```

```python
import numpy as np

# Display the correct combination
correct_combination = np.array([ALPHABET.index(c) for c in TARGET_SENTENCE])
display_board(correct_combination)

# Create a dictionnary counting the number of each char in the solution
correct_char_count = {
    char: np.count_nonzero(correct_combination == char)
    for char in np.unique(correct_combination)
}

# Print the count of each char
print(correct_char_count)

print("\nThe chromosome for the sentence '", TARGET_SENTENCE, "' is :\n", correct_combination)
```

## <font color="brown">🏋️&nbsp;&nbsp;Define fitness function</font>

```python
# Score for guessing exactly a peg
SCORE_CORRECT_PEGS = 5

# Score for guessing correctly a char
SCORE_CORRECT_CHAR = 1

# The maximum score (score of the correct combination).
# Note that for a peg guessed exactly, we score points both for
# guessing exactly and for having the right color.
MAX_SCORE = SCORE_CORRECT_PEGS * N_PEGS + SCORE_CORRECT_CHAR * N_PEGS

def score_chromosome(chromosome):
    # Char score
    chars_score = 0

    for char, char_count in correct_char_count.items():
        # We win points for each char guessed correctly
        # but not more than the number of pegs of this char
        # in the correct combination
        chars_score += min(np.count_nonzero(chromosome == char), char_count)

    chars_score *= SCORE_CORRECT_CHAR

    # Correct pegs score
    correct_peg_score = np.count_nonzero(np.array(chromosome) == correct_combination)
    correct_peg_score *= SCORE_CORRECT_PEGS

    return chars_score + correct_peg_score

def fitness_function(ga_instance, solution, solution_idx):
    return score_chromosome(solution)
```

## <font color="brown">☎️&nbsp;&nbsp;Define Callbacks </font>

```python
def on_generation(ga_instance):
    solution, solution_fitness, solution_idx = ga_instance.best_solution()
    print("Generation: ", ga_instance.generations_completed, ". Fitness: ", solution_fitness)
    solution_image = display_board(solution)
```

## <font color="brown">🚀&nbsp;&nbsp;Run genetic algorithm</font>

```python
#%pip install pygad
```

```python
import pygad

ga_instance = pygad.GA(
    sol_per_pop=100,
    num_genes=N_PEGS,
    num_generations=5000,
    num_parents_mating=50,
    fitness_func=fitness_function,
    gene_type=int,
    init_range_low=0,
    init_range_high=(N_CHARS - 1),
    on_generation=on_generation,
    mutation_type="random",
    mutation_probability=0.10,
    mutation_by_replacement=True,
    random_mutation_min_val=0.0,
    random_mutation_max_val=N_CHARS,
    crossover_type="uniform",
    crossover_probability=0.0,
    parent_selection_type="rws", # roulette wheel selection
    stop_criteria=f"reach_{MAX_SCORE}"
)

ga_instance.run()
```
